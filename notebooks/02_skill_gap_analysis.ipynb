
# CareerPrep AI — Skill-Gap Analysis

## Objective
Identify the skills a student currently has, compare them with the skills required for a chosen career, and recommend areas for improvement.

## Feature workflow
1. Select a target career.
2. Record the student's current skill levels.
3. Compare current skills with career requirements.
4. Calculate skill gaps.
5. Recommend learning priorities.

## Technology
- Python
- Jupyter Notebook
- pandas (for structured data)
- scikit-learn (for future machine learning features)
      

```python

# CareerPrep AI
# Step 1: Define skills required for each career

career_requirements = {
    "Python Developer": {
        "Python": 80,
        "SQL": 60,
        "Data Structures": 70,
        "Git and GitHub": 60,
        "Problem Solving": 75
    },

    "Full Stack Developer": {
        "HTML and CSS": 80,
        "JavaScript": 80,
        "Python": 60,
        "SQL": 65,
        "Git and GitHub": 60
    },

    "Data Scientist": {
        "Python": 80,
        "SQL": 70,
        "Statistics": 80,
        "Machine Learning": 75,
        "Data Visualization": 70
    }
}

print("Available career paths:")

for career in career_requirements:
    print("-", career)
```

```python
# CareerPrep AI
# Step 2: Record student profile and current skills

student_profile = {
    "name": "Shivangi",
    "target_career": "Python Developer",
    "current_skills": {
        "Python": 60,
        "SQL": 40,
        "Data Structures": 30,
        "Git and GitHub": 50,
        "Problem Solving": 50
    }
}

print(f"Student Name: {student_profile['name']}")
print(f"Target Career: {student_profile['target_career']}\n")

print("Current Skill Levels:")
for skill, score in student_profile["current_skills"].items():
    print(f"- {skill}: {score}%")
```

```python
def analyze_skill_gap(student_profile, career_requirements):
    target_career = student_profile["target_career"]
    required_skills = career_requirements.get(target_career, {})
    current_skills = student_profile["current_skills"]

    gaps = {}
    for skill, target_score in required_skills.items():
        current_score = current_skills.get(skill, 0)
        gaps[skill] = max(0, target_score - current_score)

    return gaps
```

```python
# CareerPrep AI
# Step 24: Test dynamic skill-gap analysis for different students

# Student 1: Shivangi
shivangi_profile = {
    "name": "Shivangi",
    "target_career": "Python Developer",
    "current_skills": {
        "Python": 60,
        "SQL": 40,
        "Data Structures": 30,
        "Git and GitHub": 50,
        "Problem Solving": 50
    }
}

# Student 2: Khushi
khushi_profile = {
    "name": "Khushi",
    "target_career": "Python Developer",
    "current_skills": {
        "Python": 80,
        "SQL": 20,
        "Data Structures": 60,
        "Git and GitHub": 70,
        "Problem Solving": 40
    }
}

def print_student_gaps(student):
    gaps = analyze_skill_gap(student, career_requirements)
    print(f"=== Skill Gap Analysis for {student['name']} ({student['target_career']}) ===")
    for skill, gap in gaps.items():
        current = student['current_skills'].get(skill, 0)
        target = career_requirements[student['target_career']][skill]
        print(f"- {skill}: Current = {current}%, Target = {target}%, Gap = {gap}%")
    print("\n")

# Run analysis for both students
print_student_gaps(shivangi_profile)
print_student_gaps(khushi_profile)
```

```python
# CareerPrep AI
# Step 5: Generate learning recommendations based on skill gaps

def generate_recommendations(student_profile, career_requirements):
    gaps = analyze_skill_gap(student_profile, career_requirements)
    
    # Sort skills by gap size in descending order (highest gap first)
    sorted_gaps = sorted(gaps.items(), key=lambda item: item[1], reverse=True)
    
    print(f"==================================================")
    print(f" CAREER RECOMMENDATION REPORT: {student_profile['name'].upper()}")
    print(f" Target Career: {student_profile['target_career']}")
    print(f"==================================================\n")
    
    top_priorities = [skill for skill, gap in sorted_gaps if gap > 0][:2]
    strengths = [skill for skill, gap in sorted_gaps if gap == 0]
    
    if strengths:
        print("🎯 Key Strengths (Requirement Met):")
        for skill in strengths:
            print(f"  - {skill}")
        print()
        
    print("🚀 Top Learning Priorities:")
    for skill, gap in sorted_gaps:
        if gap > 0:
            print(f"  - {skill}: Need +{gap}% improvement")
            
    print("\n💡 Suggested Next Step:")
    if top_priorities:
        print(f"  Focus immediately on improving: {', '.join(top_priorities)}.")
    else:
        print("  You meet all requirements for this career path! Ready for advanced projects.")
    print("\n")

# Run recommendations for Shivangi and Khushi
generate_recommendations(shivangi_profile, career_requirements)
generate_recommendations(khushi_profile, career_requirements)
```
