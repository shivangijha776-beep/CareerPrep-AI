# CareerPrep AI — Advanced Skill Extraction System

## Objective
To improve the existing job description analysis system by identifying skills even when they are expressed using different words or phrases.

## Problem with the Previous System
The previous system used exact keyword matching. This could miss skills when a job description used alternative expressions.

For example:
- Python programming → Python
- Relational database knowledge → SQL
- Version control experience → Git and GitHub

## Proposed Solution
Create a skill dictionary containing standard skill names and their synonyms.

The system will:
1. Accept a job description.
2. Process its text.
3. Search for known skill names and synonyms.
4. Map recognised expressions to standard skill names.
5. Generate a structured list of extracted skills.

## Technologies
- Python
- Pandas
- Natural Language Processing (NLP)
- Regular Expressions (re)

## Expected Outcome
A more flexible skill extraction system that can recognise different expressions of the same technical skill.

```python
skill_dictionary = {
    "Python": [
        "python",
        "python programming",
        "python development"
    ],

    "SQL": [
        "sql",
        "structured query language",
        "relational database queries"
    ],

    "Data Structures": [
        "data structures",
        "arrays",
        "linked lists",
        "stacks",
        "queues"
    ],

    "Algorithms": [
        "algorithms",
        "algorithm design",
        "sorting and searching"
    ],

    "Machine Learning": [
        "machine learning",
        "ml",
        "predictive modelling",
        "predictive modeling"
    ],

    "Git and GitHub": [
        "git",
        "github",
        "version control",
        "git and github"
    ],

    "Object-Oriented Programming": [
        "object-oriented programming",
        "oops",
        "oop",
        "classes and objects"
    ],

    "Data Analysis": [
        "data analysis",
        "data analytics",
        "data cleaning"
    ],

    "Pandas": [
        "pandas",
        "python pandas"
    ],

    "NumPy": [
        "numpy",
        "numerical python"
    ]
}

print("Skill Dictionary Created Successfully!")

for skill, synonyms in skill_dictionary.items():
    print(skill, ":", synonyms)
```

```python
import re

print("Regular Expressions library imported successfully!")
```

```python
def extract_skills(text, skill_dictionary):

    text = text.lower()
    extracted_skills = []

    for skill, synonyms in skill_dictionary.items():

        for synonym in synonyms:

            pattern = r"\b" + re.escape(synonym.lower()) + r"\b"

            if re.search(pattern, text):
                extracted_skills.append(skill)
                break

    return extracted_skills


print("Advanced Skill Extraction Function Created!")
```

```python
job_description = """
We are hiring a Python Developer.

The candidate should have experience in Python programming,
relational database queries, arrays, linked lists,
algorithm design, predictive modelling, version control,
and object-oriented programming.

Knowledge of data analytics, numerical Python,
and Python pandas is also required.
"""

print("Sample Job Description:")
print(job_description)
```

```python
extracted_skills = extract_skills(job_description, skill_dictionary)

print("EXTRACTED SKILLS")
print("-------------------------")

for skill in extracted_skills:
    print("-", skill)

print("\nTotal Skills Identified:", len(extracted_skills))
```

```python
old_skill_database = [
    "Python",
    "SQL",
    "Data Structures",
    "Algorithms",
    "Machine Learning",
    "Git and GitHub",
    "Object-Oriented Programming",
    "Data Analysis",
    "Pandas",
    "NumPy"
]

old_extracted_skills = []

job_text = job_description.lower()

for skill in old_skill_database:
    if skill.lower() in job_text:
        old_extracted_skills.append(skill)

print("OLD METHOD - Exact Keyword Matching")
print(old_extracted_skills)
print("Skills Found:", len(old_extracted_skills))

print("\nNEW METHOD - Synonym-Based Extraction")
print(extracted_skills)
print("Skills Found:", len(extracted_skills))
```

```python
expected_skills = [
    "Python",
    "SQL",
    "Data Structures",
    "Algorithms",
    "Machine Learning",
    "Git and GitHub",
    "Object-Oriented Programming",
    "Data Analysis",
    "Pandas",
    "NumPy"
]

correctly_identified = [
    skill for skill in expected_skills
    if skill in extracted_skills
]

extraction_percentage = (
    len(correctly_identified) / len(expected_skills)
) * 100

print("SKILL EXTRACTION EVALUATION")
print("---------------------------")
print("Expected Skills:", len(expected_skills))
print("Identified Skills:", len(correctly_identified))
print("Extraction Coverage:", round(extraction_percentage, 2), "%")
```

```python
student_profile = {
    "name": "Shivangi",
    "skills": [
        "Python",
        "SQL",
        "Git and GitHub",
        "Problem Solving",
        "Pandas",
        "NumPy"
    ]
}

print("STUDENT PROFILE")
print("-------------------------")
print("Name:", student_profile["name"])
print("Skills:", student_profile["skills"])
print("Total Skills:", len(student_profile["skills"]))
```

```python
student_skills = student_profile["skills"]

matched_skills = []
missing_skills = []

for skill in extracted_skills:
    if skill in student_skills:
        matched_skills.append(skill)
    else:
        missing_skills.append(skill)

print("STUDENT-JOB SKILL COMPARISON")
print("----------------------------")

print("\nMATCHED SKILLS:")
for skill in matched_skills:
    print("-", skill)

print("\nMISSING SKILLS:")
for skill in missing_skills:
    print("-", skill)

print("\nTotal Required Skills:", len(extracted_skills))
print("Matched Skills:", len(matched_skills))
print("Missing Skills:", len(missing_skills))
```

```python
total_required = len(extracted_skills)
total_matched = len(matched_skills)

if total_required > 0:
    job_match_percentage = (total_matched / total_required) * 100
else:
    job_match_percentage = 0

print("ADVANCED JOB MATCH REPORT")
print("-------------------------")
print("Student:", student_profile["name"])
print("Required Skills:", total_required)
print("Matched Skills:", total_matched)
print("Missing Skills:", len(missing_skills))
print("Job Match Percentage:", round(job_match_percentage, 2), "%")
```

```python
improvement_plan = []

for skill in missing_skills:
    improvement_plan.append({
        "Skill": skill,
        "Status": "Needs Improvement",
        "Action": "Learn the basics and practise related problems"
    })

print("PERSONALISED IMPROVEMENT PLAN")
print("-----------------------------")

for item in improvement_plan:
    print("Skill:", item["Skill"])
    print("Status:", item["Status"])
    print("Action:", item["Action"])
    print()
```

```python
import matplotlib.pyplot as plt

skill_status = ["Matched Skills", "Missing Skills"]
skill_counts = [len(matched_skills), len(missing_skills)]

plt.figure(figsize=(7, 5))
plt.bar(skill_status, skill_counts)

plt.title("Student Skill Gap Analysis")
plt.xlabel("Skill Status")
plt.ylabel("Number of Skills")
plt.ylim(0, total_required)

plt.show()
```
